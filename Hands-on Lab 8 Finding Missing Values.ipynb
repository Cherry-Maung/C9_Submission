### 3. Finding Missing Values
##### Task 2: Identify missing values for all columns.


```python
## Write your code here
df.isnull().sum()
```

```python
df.isna()
```

##### Task 3: Visualize missing values using a heatmap (Using seaborn library).



```python
plt.figure(figsize=(10, 6)) # 2. Visualize missing values using a Seaborn heatmap
sns.heatmap(df.isnull(), cbar=False, cmap='viridis')
plt.title('Heatmap of Missing Values')
plt.show()
```

##### Task 4: Count the number of missing rows for a specific column (e.g., `Employment`).


```python
## Write your code here
count_mv_emp = df['Employment'].isnull().sum()
print(f"Number of missing rows in the 'Employment' column: {count_mv_emp}")
```

```python
## Write your code here
count_mv_remotework = df['RemoteWork'].isnull().sum()
print(f"Number of missing rows in the 'RemoteWork' column: {count_mv_remotework}")
```

### 4. Imputing Missing Values
##### Task 5: Identify the most frequent (majority) value in a specific column (e.g., `Employment`).


```python
count = df['RemoteWork'].value_counts()
```

```python
or_rw = df[['RemoteWork']]
```

```python
## Write your code here
RemoteWork_md = df['RemoteWork'].mode()[0]
print(f"The most frequent value of Remote Work is {RemoteWork_md}")
```

```python
# Returns just the value
most_frequent = df['RemoteWork'].value_counts().idxmax()
most_frequent
```

##### Task 6: Impute missing values in the `Employment` column with the most frequent value.



```python
## Write your code here
df['RemoteWork'] = df['RemoteWork'].fillna(RemoteWork_md)
```

```python
count_mv_remotework = df['RemoteWork'].isnull().sum()
print(f"Number of missing rows in the 'RemoteWork' column: {count_mv_remotework}")
```

### 5. Visualizing Imputed Data
##### Task 7: Visualize the distribution of a column after imputation (e.g., `Employment`).


```python
import matplotlib.pyplot as plt
import seaborn as sns
# Set up 1 row, 2 columns for Comparing Original vs. Imputed
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
# Plot on first subplot
sns.histplot(or_rw['RemoteWork'], ax=axes[0]).set_title('Original Data')
# Plot on second subplot
sns.histplot(df['RemoteWork'], ax=axes[1]).set_title('Imputed Data')
#sns.set_context("notebook", font_scale=0.7)
plt.tight_layout() # Optimize spacing
```
