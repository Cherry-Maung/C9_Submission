### Step 1: Import Required Libraries


```python
import pandas as pd
```

### Step 2: Load the Dataset Into a Dataframe


#### **Read Data**
<p>
The functions below will download the dataset into your browser:
</p>


```python
file_path = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/n01PQ9pSmiRX6520flujwQ/survey-data.csv"
df = pd.read_csv(file_path)

# Display the first few rows to ensure it loaded correctly
print(df.head())
```

### Step 3. Finding and Removing Duplicates
##### Task 1: Identify duplicate rows in the dataset.


```python
## Write your code here
duplicate_rows_mask = df.duplicated(keep=False)
df_duplicates_all = df[duplicate_rows_mask]
print(f'Number of duplicate rows: {df_duplicates_all.shape[0]}')
```

##### Task 2: Remove the duplicate rows from the dataframe.



```python
## Write your code here
df.drop_duplicates(inplace=True)
df.shape
```

### Step 4: Finding Missing Values
##### Task 3: Find the missing values for all columns.


```python
missing_data = df.isnull()

for column in missing_data.columns.values.tolist():
    print(column)
    print (missing_data[column].value_counts())
    print("")
```

```python
## Write your code here
df.isnull().sum()
```

##### Task 4: Find out how many rows are missing in the column RemoteWork.


```python
## Write your code here
mv_rmwk = df['RemoteWork'].isnull().sum()
print(f"Number of missing rows in the 'RemoteWork' column: {mv_rmwk}")
```

### Step 5. Imputing Missing Values
##### Task 5: Find the value counts for the column RemoteWork.


```python
## Write your code here
vc_rmwk = df['RemoteWork'].value_counts()
print(f"Count the value in the 'RemoteWork' column: {vc_rmwk}")
```

##### Task 6: Identify the most frequent (majority) value in the RemoteWork column.



```python
## Write your code here
most_frequent = df['RemoteWork'].value_counts().idxmax()
most_frequent
```

##### Task 7: Impute (replace) all the empty rows in the column RemoteWork with the majority value.



```python
## Write your code here
df['RemoteWork'] = df['RemoteWork'].fillna(most_frequent)
```

```python
count_mv_remotework = df['RemoteWork'].isnull().sum()
count_mv_remotework
```

##### Task 8: Check for any compensation-related columns and describe their distribution.



```python
## Write your code here
print(f'Number of Missing Value for Yearly Compensation column: {df['ConvertedCompYearly'].isnull().sum()}')
```

```python
mean_comp = df['ConvertedCompYearly'].mean()
print(mean_comp)
```

```python
df['ConvertedCompYearly_FMV'] = df['ConvertedCompYearly'].fillna(mean_comp)
```

```python
print(f'Number of Missing Value for Yearly Compensation column: {df['ConvertedCompYearly_FMV'].isnull().sum()}')
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
scaler = MinMaxScaler()
df['ConvertedCompYearly_Normalized'] = scaler.fit_transform(df[['ConvertedCompYearly_FMV']])
```

```python
df['ConvertedCompYearly_Normalized'].loc[30000]
```

```python
df['ConvertedCompYearly_Normalized'].describe()
```

```python
print(f'Number of Missing Value for Compensation Total column: {df['CompTotal'].isnull().sum()}')
```

```python
mean_compt = df['CompTotal'].mean()
print(mean_compt)
```

```python
df['CompTotal_FMCT'] = df['CompTotal'].fillna(mean_compt)
```

```python
print(f'Number of Missing Value for Compensation Total column: {df['CompTotal_FMCT'].isnull().sum()}')
```

```python
df[['CompTotal_FMCT','ConvertedCompYearly_Normalized']].describe()
```
